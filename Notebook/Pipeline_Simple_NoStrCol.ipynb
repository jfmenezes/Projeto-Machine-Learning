```python

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import joblib

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, f1_score

RANDOM_STATE = 42
PLOT_BLUE, PLOT_ORANGE = "#1779A8", "#C96A00"  # course plot colours

pd.set_option('display.max_columns', 50) # tells pandas to display up to 50 columns when printing a DataFrame
pd.set_option('display.max_rows', 50)    # tells pandas to display up to 50 rows when printing a DataFrame
```

```python
Directory = '/Users/josemenezes/Faculdade/Projetos/Projeto ML/Projeto-Machine-Learning/project/data/'

train_data = pd.read_csv(Directory + 'train.csv', index_col='customer_id')
teste_data = pd.read_csv(Directory + 'test.csv',index_col='customer_id')
clients = pd.read_csv(Directory + 'clientes.csv', index_col='customer_id')

print("train data:", train_data.shape, "| test data:", teste_data.shape, "| clients data:", clients.shape)
```

```python
merged_data = train_data.merge(clients, on='customer_id', how='inner')
```

```python
merged_data.head()
```

```python
merged_data=merged_data.drop('customer_type', axis=1)
```

```python
merged_data.info()
```

```python
for col in merged_data.columns: # I used this for loop to remove all str columns to test a first model
    if merged_data[col].dtype == "str":
        merged_data = merged_data.drop(columns=col)
```

```python
merged_data.info()
```

```python
merged_data=merged_data.drop('random_noise', axis=1)
```

```python
X = merged_data.drop(columns='repeat_purchase_90d')
y = merged_data['repeat_purchase_90d']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
X,
y, 
train_size=0.7,
stratify=y,
random_state=RANDOM_STATE,
)
len(X_train), len(X_test)
```

```python
tree_model = DecisionTreeClassifier(random_state=RANDOM_STATE)
tree_model.fit(X_train, y_train)
```

```python
predictions_train=tree_model.predict(X_train)
predictions_test=tree_model.predict(X_test)
predictions_test[:10]
```

```python
unt_train_acc = tree_model.score(X_train, y_train)
unt_test_acc = tree_model.score(X_test, y_test)
unt_train_f1 = f1_score(y_train, predictions_train)
unt_test_f1 = f1_score(y_test, predictions_test)

print(f'Default tree train: accuracy = {unt_train_acc:.3f} F1 = {unt_train_f1:.3f}')
print(f'Default tree test: accuracy = {unt_test_acc:.3f} F1 = {unt_test_f1:.3f}')
```
